# Görüntüden Renk Tanıma

Bu notebook, bir görüntüdeki piksellerin rengini tanımak için kullanılır. Notebook, renkleri bir CSV dosyasından yükler, bir görüntüdeki pikselin rengini tanır ve sonucu yazdırır.

## 1. Kütüphanelerin Yüklenmesi

Bu bölümde, projemizde kullanacağımız temel kütüphaneleri yüklüyoruz. `cv2` (OpenCV) görüntü işleme için, `numpy` sayısal operasyonlar için, `pandas` veri işleme ve CSV dosyalarını okumak için ve `os` işletim sistemi ile ilgili işlemler için kullanılır.

```python
import cv2
import numpy as np
import pandas as pd
import os
```

## 2. Yardımcı Fonksiyonlar

Bu bölümde, renk tanıma işlemi için gerekli olan yardımcı fonksiyonları tanımlıyoruz.

### `load_color_lookup_table` Fonksiyonu

Bu fonksiyon, renklerin isimlerini ve RGB değerlerini içeren CSV dosyasını okur. `colors.csv` dosyasını okur ve renk adı ile RGB değerini içeren bir sözlük döndürür.

```python
def load_color_lookup_table(path='colors.csv'):
    color_df = pd.read_csv(path, names=['color_name_id', 'color_name', 'hex', 'R', 'G', 'B'])
    defined_colors_dict = {row['color_name']: np.array([row['R'], row['G'], row['B']])
                           for index, row in color_df.iterrows()}
    return defined_colors_dict
```

### `recognize_color` Fonksiyonu

Bu fonksiyon, bir RGB piksel değeri ve tanımlı renklerin bir sözlüğünü alır. Pikselin rengini, tanımlı renkler arasındaki en yakın rengi bularak tanır. En yakın rengi bulmak için Öklid mesafesini kullanır.

```python
def recognize_color(rgb_pixel, defined_colors_dict):
    min_distance = float('inf')
    closest_color_name = "Bilinmeyen"

    pixel_array = np.array(rgb_pixel)

    for color_name, color_rgb_array in defined_colors_dict.items():
        distance = np.linalg.norm(pixel_array - color_rgb_array)
        
        if distance < min_distance:
            min_distance = distance
            closest_color_name = color_name
            
    return closest_color_name, min_distance
```

## 3. Renklerin ve Görüntünün Yüklenmesi

Bu bölümde, `load_color_lookup_table` fonksiyonunu kullanarak renkleri yüklüyoruz ve `cv2.imread` fonksiyonunu kullanarak bir görüntü okuyoruz. Görüntü BGR formatında okunur, bu yüzden onu RGB formatına dönüştürüyoruz.

```python
defined_colors = load_color_lookup_table()
img_bgr = cv2.imread('data/images/colour.jpeg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
```

## 4. Bir Piksel Seçin ve Rengini Tanıyın

Bu bölümde, görüntüden bir piksel seçiyoruz (örneğin, koordinat (100, 100)'deki piksel) ve `recognize_color` fonksiyonunu kullanarak rengini tanıyoruz. Son olarak, seçilen koordinatları, pikselin RGB değerlerini ve tanınan renk adını yazdırıyoruz.

```python
x_coord = 150
y_coord = 100
selected_pixel_rgb = img_rgb[y_coord, x_coord]
color_name, distance = recognize_color(selected_pixel_rgb, defined_colors)

print(f"Secilen Koordinat: ({x_coord}, {y_coord})")
print(f"RGB Degerleri: {selected_pixel_rgb}")
print(f"Taninan Renk Adi: {color_name}")
```
